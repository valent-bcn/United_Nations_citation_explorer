```python
# reset index
treaties.reset_index(drop=True, inplace=True)

# remove the last point of column note, if it exists, and strip whitespace
treaties['note'] = treaties['note'].str.strip()
treaties['note'] = treaties['note'].str.strip('.').str.strip()

# remove the last point of column summary, if it exists, and strip whitespace
treaties['summary'] = treaties['summary'].str.strip()
treaties['summary'] = treaties['summary'].str.strip('.').str.strip()

treaties.to_csv('./wiki-treaties.csv', index=False)
```

```python
import pandas as pd

df = pd.read_csv('wikipedia_treaties-all.csv')
df.head()

# Keep only rows where 'year' is fully numeric
treaties = df[df['year'].astype(str).str.isdigit()].copy()

treaties['year'] = treaties['year'].astype(int)
treaties = treaties[treaties['year'].isin(range(1600, 1900))]

# reset index
treaties.reset_index(drop=True, inplace=True)

# remove the last point of column note, if it exists, and strip whitespace
treaties['note'] = treaties['note'].str.strip()
treaties['note'] = treaties['note'].str.strip('.').str.strip()

# remove the last point of column summary, if it exists, and strip whitespace
treaties['summary'] = treaties['summary'].str.strip()
treaties['summary'] = treaties['summary'].str.strip('.').str.strip()

treaties.to_csv('./wiki-treaties_pre-1900.csv', index=False)
```
