```python
import pandas as pd

df = pd.read_csv("spiritRAG_corpus.csv")
df_res = pd.read_csv("../resolutions/ga_resolutions_1946_2019.csv")

df
```

```python
"""
SpiritRAG corpora contains documents that are already present in the collection of
Mesquita and Pires. Meaning that it is not necessarily to concatenate both datasets.
The Mesquita and Pires corpora encompasses resolution until 16 September 2019.
Later than are General Assembly resolution are highly valuable. Furthermore, the
SpiritRAG corpora includes documents outside the General Assembly, they are valuable
independently of the date, since no records of such documents are presented by Mesquita.
"""

df["publication_date"] = pd.to_datetime(
    df["publication_date"],
    format="%d/%m/%Y",
    errors="coerce"
)

cutoff = pd.Timestamp("2019-09-16")

df = df[
    ~(df["symbol"].str.startswith("A/RES", na=False) & (df["publication_date"] < cutoff))
]

df["publication_date"] = df["publication_date"].dt.strftime("%d %B %Y").str.lower()
print(f"Found {len(df)} rows")
```

```python
df.rename(columns={"symbol": "res_id2", "publication_date": "date_c"}, inplace=True)
```

```python
cols = ["res_id2", "date_c", "content"]

df_1946_2025 = pd.concat(
    [df[cols], df_res[cols]],
    ignore_index=True
)

df_1946_2025.to_csv("../resolutions/ga_resolutions_1946_2025.csv", index=False)
print(f"Saved {len(df_1946_2025)} rows")
```
